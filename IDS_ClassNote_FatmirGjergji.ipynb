## Control Flows

## What are Control Flows?
Control flows are the decision-making structures in programming that determine the execution path of a program based on conditions or loops. They direct the computer on when to start, stop, or skip specific instructions based on logical decisions or repetitive conditions.

## Key Concepts in Control Flows

### 1. Conditional Statements
   - **If Statements**: Execute a block of code only if a specified condition is true.
   - **Else Statements**: Execute a block of code if the condition in the if statement is false.
   - **Elif Statements**: Specify a new condition to test if the previous conditions were false.

### 2. Looping Statements
   - **For Loops**: Iterate over a sequence (like a list, tuple, or string) and execute a block of code once for each item.
   - **While Loops**: Continuously execute a block of code as long as a condition is true.

### 3. Control Statements
   - **Break**: Exit the loop entirely.
   - **Continue**: Skip the rest of the code inside the loop for the current iteration and move to the next iteration.
   - **Pass**: A placeholder statement, syntactically needed, but performs no operation.

#quiz string , manipulation , regular expression

The following is a table of the repetition markers available for use in regular expressions:

| Character | Description | Example |
|-----------|-------------|---------|
| ``?`` | Match zero or one repetitions of preceding  | ``"ab?"`` matches ``"a"`` or ``"ab"`` |
| ``*`` | Match zero or more repetitions of preceding | ``"ab*"`` matches ``"a"``, ``"ab"``, ``"abb"``, ``"abbb"``... |
| ``+`` | Match one or more repetitions of preceding  | ``"ab+"`` matches ``"ab"``, ``"abb"``, ``"abbb"``... but not ``"a"`` |
| ``{n}`` | Match ``n`` repetitions of preeeding | ``"ab{2}"`` matches ``"abb"`` |
| ``{m,n}`` | Match between ``m`` and ``n`` repetitions of preceding | ``"ab{2,3}"`` matches ``"abb"`` or ``"abbb"`` |

# Numpy
NumPy is a library for the Python programming language, adding support for large, multi-dimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays. It is one of the core packages for scientific computing with Python and is widely used in machine learning, data science, and scientific computing communities.

NumPy provides efficient and convenient support for array computation, and its multi-dimensional arrays provide a structure for organizing and processing large amounts of homogeneous data, making it a powerful tool for solving a wide range of computational problems in science and engineering.


## Why numpy is important?

NumPy is a library for the Python programming language that provides support for arrays and matrices, along with a collection of mathematical functions to operate on these arrays. It is widely used for numerical computing, data analysis, and scientific computing.

There are several reasons why NumPy is important:

- Efficient Computation: NumPy arrays are much more efficient for numerical computations compared to lists in Python, as the data is stored in a contiguous block of memory and can be processed much more quickly by the computer.

- Broadcasting: NumPy supports broadcasting, which allows for operations to be performed between arrays of different shapes, by repeating the smaller array to match the shape of the larger array.

- Integration with Other Libraries: NumPy provides support for reading and writing arrays to and from files, as well as integrating with other libraries, such as Matplotlib for plotting and visualization, and Pandas for data analysis.

- Element-wise Operations: One of the key features of NumPy is its ability to perform element-wise operations on arrays, allowing for fast and efficient computation.

# Introduction
Pandas is a popular open-source Python library used for data manipulation and analysis. It provides easy-to-use data structures for storing and manipulating data, including the DataFrame, Series, and Panel objects.

The DataFrame is perhaps the most commonly used object in Pandas. It is a two-dimensional table-like structure with columns of potentially different data types, similar to a spreadsheet or SQL table. The Series object, on the other hand, is a one-dimensional labeled array that can hold any data type.

Pandas is widely used in data science, machine learning, and scientific computing to clean, transform, and analyze data. It can handle a variety of data sources, including CSV, Excel, SQL databases, and web APIs. Pandas also integrates well with other Python libraries like NumPy, SciPy, and Matplotlib.


# Why Pandas is import?

Pandas is an important tool in data analysis and manipulation for several reasons:

Data Manipulation: Pandas provides a variety of data structures and functions for manipulating data. This includes filtering, grouping, sorting, joining, and aggregating data. These functions help to clean and transform data into a format that can be easily analyzed.

Data Cleaning: Pandas provides tools for cleaning and handling missing data, which is a common issue in real-world data. It also provides functions for data normalization, formatting, and data type conversions.

Data Analysis: Pandas makes it easy to perform data analysis tasks such as statistical analysis, data visualization, and data exploration. It has built-in functions for calculating descriptive statistics, correlation, and regression analysis. It also integrates well with other libraries for data visualization like Matplotlib and Seaborn.

Data Integration: Pandas can integrate with a variety of data sources including SQL databases, CSV files, and Excel spreadsheets. This makes it easy to read, write, and combine data from different sources.

Large Dataset Handling: Pandas is optimized for handling large datasets efficiently. It can handle millions of rows and columns, which is crucial for data analysis in today's big data world.

# Pandas' Datatype
Pandas provides several data types that are commonly used for data manipulation and analysis. The most common data types in Pandas are:

- Series: A one-dimensional array-like object that can hold any data type, such as integers, floats, strings, or even Python objects.

- DataFrame: A two-dimensional table-like structure that can hold heterogeneous data types, similar to a spreadsheet or SQL table.

- Index: A one-dimensional array that labels the rows or columns of a DataFrame or Series.

- DateTime: A data type that represents dates and times.

- Categorical: A data type that represents categorical data with a fixed set of categories.

- Boolean: A data type that represents boolean values (True or False).

- Numeric: A data type that represents numeric data, including integers, floats, and complex numbers.

- Object: A data type that represents any Python object, including strings and custom data types.

Pandas also provides support for several other data types, including interval data, sparse data, and mixed data types. The flexibility of Pandas data types makes it easy to work with a wide variety of data sources and perform complex data manipulation tasks.

you're instructing pandas to drop the column labeled `'new'` from the DataFrame. (If you set `axis=0`, it would instead drop rows with the index label `'new'`.)

````python
import requests
import pandas as pd
from bs4 import BeautifulSoup

# Initialize an empty dataframe
df_positions = pd.DataFrame(columns=[
    'Institution Name', 'Position ID', 'Position Title',
    'Position Location', 'Subject Area', 'Application Deadline', 'Apply Link'
])

# URL of the job listings page
url = "https://www.mathjobs.org/jobs/list"

# Send a request to fetch the HTML content
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Find all job entries
job_entries = soup.find_all('div', class_='clr')

# Loop through each job entry and extract data
for job in job_entries:
    institution_name = job.find('h3', class_='x1')
    if institution_name:
        institution_name = institution_name.get_text(strip=True)
    else:
        institution_name = "Not Available"

    position_tag = job.find('a')
    if position_tag:
        position_title = position_tag.get_text(strip=True)
        position_id = position_tag['href'].split('/')[-1]
    else:
        position_title = "Not Available"
        position_id = "Not Available"

    subject_area_tag = job.find('ol', class_='sp5')
    if subject_area_tag:
        subject_area = subject_area_tag.get_text(strip=True)
    else:
        subject_area = "Not Available"

    apply_link_tag = job.find('a', href=True, text="Apply")
    if apply_link_tag:
        apply_link = apply_link_tag['href']
    else:
        apply_link = "Not Available"

    # Append the extracted data to the dataframe
    df_positions.loc[len(df_positions)] = [
        institution_name, position_id, position_title,
        "Not Available", subject_area, "Not Available", apply_link
    ]

# Display the dataframe
df_positions
````

### boxplot and violinplot

Box plots and violin plots are both used to represent the distribution of a continuous variable. However, they have some differences in the way they depict the data.

A box plot displays the median, quartiles, and any outliers in the data using a rectangular box and whiskers. The rectangular box represents the middle 50% of the data, with the median indicated by a line in the box. The whiskers extend from the box to the highest and lowest values within 1.5 times the interquartile range (IQR) of the data, and any data points outside of the whiskers are considered outliers. Box plots are useful for quickly summarizing the spread of the data and identifying any potential outliers.

On the other hand, a violin plot displays the density of the data at different values using a kernel density plot. The width of the plot at a given point represents the density of the data at that point, with the maximum width indicating the mode of the distribution. The white dot in the center of the plot represents the median. Violin plots are useful for showing the shape of the distribution and providing more information about the density of the data.

In summary, box plots are useful for summarizing the spread of the data and identifying outliers, while violin plots provide more information about the shape and density of the data. The choice between these two types of plots will depend on the specific needs of the analysis and the nature of the data being represented.

### stripplot and swarmplot

**Inflation adjustment**

To calculate the equivalent value of one dollar in 1912 in terms of its purchasing power in 2025, you can use the **consumer price index (CPI)** to adjust for inflation. The CPI measures the average change over time in the prices paid by urban consumers for a market basket of consumer goods and services. Here's how you can do it:

- Find the CPI for 1912 and 2025.
- Divide the CPI for 2025 by the CPI for 1912 to get the **inflation factor**.
- Multiply one dollar by the inflation factor to find its equivalent value in 2025.

````python
url = 'https://www.minneapolisfed.org/about-us/monetary-policy/inflation-calculator/consumer-price-index-1913-'
df_inflation = pd.read_html(url)[0]
starting_year = 1913
ending_year = 2025
cpi_initial_year = 9.8 #1912
cpi_target_year = 314.4 #2025
inflation_factor = cpi_target_year/cpi_initial_year
inflation_factor
````

What is a Mosaic Plot?
 - A mosaic plot is a graphical way of displaying multivariate categorical data.

 - It divides a rectangle into tiles; size of each tile is proportional to the frequency or count of observations.

- Useful for visualizing relationships between two or more categorical variables.



Key Features
 - Each rectangle`s area = proportion of that combination.

 - Helps to quickly spot patterns, associations, or independence between variables.

- Better suited for categorical data compared to traditional bar plots.
